```python
import pandas as pd

file_path = "C:/Users/MD Tauqueer Rizwan/Downloads/india-economic-intelligence/dataraw/Major Monetary Policy Rates.xlsx"

df = pd.read_excel(file_path, header=None)

# Keep the actual data and remove the empty first column
df = df.iloc[7:, 1:9].copy()

# Remove the numbering row
df = df.iloc[1:].copy()

df.columns = [
    "Effective_Date",
    "Bank_Rate",
    "Repo_Rate",
    "Reverse_Repo_Rate",
    "SDF_Rate",
    "MSF_Rate",
    "CRR",
    "SLR"
]

df = df.reset_index(drop=True)

# Convert date and numeric columns
df["Effective_Date"] = pd.to_datetime(df["Effective_Date"], errors="coerce")

for col in df.columns[1:]:
    df[col] = pd.to_numeric(df[col].replace("-", pd.NA), errors="coerce")

display(df.head())
```

```python
# Check the data types
print(df.dtypes)

# Check missing values
print(df.isnull().sum())

# Sort by date
df = df.sort_values("Effective_Date").reset_index(drop=True)

# Display cleaned data
display(df.head())
```

```python
repo_df = df[["Effective_Date", "Repo_Rate"]].dropna().copy()

repo_df = repo_df.sort_values("Effective_Date").reset_index(drop=True)

print("Total repo-rate records:", len(repo_df))
display(repo_df.head())
display(repo_df.tail())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

plt.plot(
    repo_df["Effective_Date"],
    repo_df["Repo_Rate"],
    marker="o"
)

plt.title("RBI Repo Rate Trend (2001–2026)")
plt.xlabel("Date")
plt.ylabel("Repo Rate (%)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("Average Repo Rate:", round(repo_df["Repo_Rate"].mean(), 2), "%")
print("Maximum Repo Rate:", repo_df["Repo_Rate"].max(), "%")
print("Minimum Repo Rate:", repo_df["Repo_Rate"].min(), "%")
print("Latest Repo Rate:", repo_df.iloc[-1]["Repo_Rate"], "%")
print("Latest Update:", repo_df.iloc[-1]["Effective_Date"].date())
```

```python
first_rate = repo_df["Repo_Rate"].iloc[0]
latest_rate = repo_df["Repo_Rate"].iloc[-1]

change = latest_rate - first_rate

print("Repo Rate in 2001:", first_rate, "%")
print("Latest Repo Rate:", latest_rate, "%")
print("Overall Change:", round(change, 2), "percentage points")
```

```python
repo_df["Rate_Change"] = repo_df["Repo_Rate"].diff()

display(repo_df.head(10))
```

```python
increases = (repo_df["Rate_Change"] > 0).sum()
decreases = (repo_df["Rate_Change"] < 0).sum()
unchanged = (repo_df["Rate_Change"] == 0).sum()

print("Rate Increases:", increases)
print("Rate Decreases:", decreases)
print("No Change:", unchanged)
print("Total Changes:", increases + decreases + unchanged)
```

```python
import matplotlib.pyplot as plt

categories = ["Increases", "Decreases", "Unchanged"]
counts = [increases, decreases, unchanged]

plt.figure(figsize=(8, 5))
plt.bar(categories, counts)

plt.title("RBI Repo Rate Changes (2001–2026)")
plt.xlabel("Type of Change")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

file_path = "C:/Users/MD Tauqueer Rizwan/Downloads/india-economic-intelligence/dataraw/Annexures_for_Press_Release_in_Excel_August2026_Annex-II.xlsx"

cpi_df = pd.read_excel(file_path)

print("Shape:", cpi_df.shape)
display(cpi_df.head(10))
```

```python
cpi_df = pd.read_excel(file_path, header=None)

cpi_df = cpi_df.iloc[4:, :].copy()

cpi_df = cpi_df.iloc[:, [1, 2, 3, 4, 5, 6, 7, 8]]

cpi_df.columns = [
    "Group_Code",
    "Category",
    "Index_Rural",
    "Index_Urban",
    "Index_Combined",
    "Inflation_Rural",
    "Inflation_Urban",
    "Inflation_Combined"
]

cpi_df = cpi_df.reset_index(drop=True)

display(cpi_df.head())
```

```python
cpi_df = cpi_df.iloc[1:].reset_index(drop=True)

cpi_df.columns = [
    "Group_Code",
    "Category",
    "Index_Rural",
    "Index_Urban",
    "Index_Combined",
    "Inflation_Rural",
    "Inflation_Urban",
    "Inflation_Combined"
]

# Convert numeric columns
numeric_cols = [
    "Index_Rural", "Index_Urban", "Index_Combined",
    "Inflation_Rural", "Inflation_Urban", "Inflation_Combined"
]

for col in numeric_cols:
    cpi_df[col] = pd.to_numeric(cpi_df[col], errors="coerce")

display(cpi_df.head())
```

```python
top_inflation = cpi_df.sort_values(
    "Inflation_Combined", ascending=False
)

display(
    top_inflation[
        ["Category", "Inflation_Combined"]
    ].head(10)
)
```

```python
import matplotlib.pyplot as plt

top10 = cpi_df.nlargest(10, "Inflation_Combined")
top10 = top10.sort_values("Inflation_Combined")

plt.figure(figsize=(10, 6))

plt.barh(
    top10["Category"],
    top10["Inflation_Combined"]
)

plt.title("Top 10 Categories by Combined Inflation — August 2026")
plt.xlabel("Inflation (%)")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

```python
lowest_inflation = cpi_df.nsmallest(10, "Inflation_Combined")

display(
    lowest_inflation[
        ["Category", "Inflation_Combined"]
    ]
)
```
